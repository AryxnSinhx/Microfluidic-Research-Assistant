```python
def search(query, k=4):
    query_emb = embed_model.encode([query], normalize_embeddings=True, convert_to_numpy=True)
    scores, indices = index.search(query_emb.astype(np.float32), k)
    results = []
    for score, idx in zip(scores[0], indices[0]):
        chunk = chunks[idx]
        results.append({
            'score': float(score),
            'chunk_id': chunk['chunk_id'],
            'paper_title': chunk['paper_title'],
            'section': chunk['section'],
            'text': chunk['text']
        })
    return results


def rag_generate(query, k=4, max_new_tokens=300, rep_penalty=1.3, ngram=3):
    retrieved = search(query, k=k)

    context = ""
    for i, r in enumerate(retrieved, 1):
        context += f"[DOC {i}] ({r['paper_title']}, {r['section']}) {r['text']}\n\n"

    prompt = (
        f"[INST] Answer the question using only the provided documents. "
        f"Cite which document(s) support your answer. If the documents don't "
        f"contain the answer, say so explicitly rather than guessing.\n\n"
        f"{context}"
        f"Question: {query} [/INST]"
    )

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=rep_penalty,
            no_repeat_ngram_size=ngram,
            pad_token_id=tokenizer.eos_token_id
        )
    return tokenizer.decode(output[0][inputs['input_ids'].shape[1]:], skip_special_tokens=True)
```

```python
query = "How does the diazo coupling reaction between bilirubin and diazotized sulfanilic acid work, and what does it produce?"
print(rag_generate(query))
```

```python
import re

def truncate_repetition(text, min_repeats=1):
    """Detect and truncate a sentence that repeats consecutively."""
    sentences = re.split(r'(?<=[.!?])\s+', text.strip())
    cleaned = []
    for i, s in enumerate(sentences):
        if i >= min_repeats and all(sentences[i-j] == s for j in range(1, min_repeats+1)):
            break  # stop at the point repetition starts
        cleaned.append(s)
    return ' '.join(cleaned)


def rag_generate(query, k=4, max_new_tokens=300, rep_penalty=1.1, score_threshold=0.4):
    retrieved = search(query, k=k)
    
    if retrieved[0]['score'] < score_threshold:
        return "No sufficiently relevant source found in the corpus for this question."
    
    context = ""
    for i, r in enumerate(retrieved, 1):
        context += f"[DOC {i}] ({r['paper_title']}, {r['section']}) {r['text']}\n\n"
    
    prompt = (
        f"[INST] Answer the question using only the provided documents. "
        f"Cite which document(s) support your answer. If the documents don't "
        f"contain the answer, say so explicitly rather than guessing.\n\n"
        f"{context}Question: {query} [/INST]"
    )
    
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        output = model.generate(
            **inputs, max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=rep_penalty,
            pad_token_id=tokenizer.eos_token_id
        )
    raw = tokenizer.decode(output[0][inputs['input_ids'].shape[1]:], skip_special_tokens=True)
    return truncate_repetition(raw)

# Test both queries at 1.1
print("=== Grounded query ===")
print(rag_generate(query))
print("\n=== Out-of-scope query ===")
print(rag_generate(out_of_scope_query))
```
